```python
from data_utils import read_conll, label_set
 
# --- Correspondance label <-> entier ---------------------------------------
# Le modele ne comprend pas "B-PER", il lui faut des entiers.
LABELS = ["O", "B-PER", "I-PER", "B-LOC", "I-LOC",
          "B-DATE", "I-DATE", "B-ORG", "I-ORG"]
label2id = {l: i for i, l in enumerate(LABELS)}
id2label = {i: l for l, i in label2id.items()}
 
 
def build_hf_dataset(path):
    """Lit un fichier CoNLL et renvoie deux listes paralleles :
    tokens_list (liste de listes de mots) et tags_list (liste de listes d'ids).
    C'est le format d'entree standard attendu par le tokenizer HF.
    """
    sentences = read_conll(path)
    tokens_list = [s["tokens"] for s in sentences]
    tags_list = [[label2id[l] for l in s["labels"]] for s in sentences]
    return tokens_list, tags_list
```
